```python
import pandas as pd
import requests

# Official San Francisco EMS dataset
url = "https://data.sf.gov/resource/nuek-vuh3.json"

# Request 1000 records
params = {
    "$limit": 1000
}

# Connect to API
response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

# Convert to DataFrame
df = pd.DataFrame(response.json())

# Explore dataset
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst records:")
display(df.head())

print("\nMissing values:")
display(df.isnull().sum())

print("\nEmergency types:")
display(df["call_type"].value_counts())
```

```python

url = "https://data.sf.gov/resource/nuek-vuh3.json"

# Retrieve just 10 records
response = requests.get(url, params={"$limit": 10})
response.raise_for_status()

df = pd.DataFrame(response.json())

# Check dataset structure
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
important_columns = [
    "incident_number",
    "call_number",
    "unit_id",
    "unit_type",
    "call_type",
    "received_dttm",
    "dispatch_dttm",
    "on_scene_dttm",
    "available_dttm",
    "case_location"
]

display(df[important_columns].head(10))
```

```python

params = {
    "$select": "count(*) as total_records",
    "$where": (
        "received_dttm >= '2025-01-01T00:00:00' "
        "AND received_dttm < '2025-02-01T00:00:00'"
    )
}

response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

result = response.json()

print("Total records in January 2025:", result[0]["total_records"])
```

```python
params = {
    "$where": (
        "received_dttm >= '2025-01-01T00:00:00' "
        "AND received_dttm < '2025-02-01T00:00:00'"
    ),
    "$order": "rowid ASC",
    "$limit": 10000,
    "$offset": 0
}

all_records = []

while True:
    response = requests.get(url, params=params, timeout=120)
    response.raise_for_status()

    batch = response.json()

    if not batch:
        break

    all_records.extend(batch)

    print(f"Retrieved {len(all_records)} records")

    if len(batch) < params["$limit"]:
        break

    params["$offset"] += len(batch)

# Convert to DataFrame
df_jan = pd.DataFrame(all_records)

print("\nFinal dataset shape:", df_jan.shape)
```

```python
print("Total response records:", len(df_jan))

print("Unique incidents:", df_jan["incident_number"].nunique())

print("Unique calls:", df_jan["call_number"].nunique())

print("Unique emergency units:", df_jan["unit_id"].nunique())
```

```python
print("Total response records:", len(df_jan))

print("Unique incidents:", df_jan["incident_number"].nunique())

print("Unique calls:", df_jan["call_number"].nunique())

print("Unique emergency units:", df_jan["unit_id"].nunique())
```

```python
# Count responses by emergency unit type
unit_types = df_jan["unit_type"].value_counts(dropna=False)

print("Response records by unit type:")
print(unit_types)

print("\nUnique unit IDs by type:")
print(df_jan.groupby("unit_type")["unit_id"].nunique())
```

```python
# Select potential ambulance resources
ambulance_types = ["MEDIC", "BLS", "PRIVATE"]

ambulances = df_jan[
    df_jan["unit_type"].isin(ambulance_types)
].copy()

# Summarise the data
summary = ambulances.groupby("unit_type").agg(
    total_responses=("unit_id", "count"),
    unique_units=("unit_id", "nunique"),
    transports=("transport_dttm", "count"),
    availability_records=("available_dttm", "count")
)

display(summary)
```

```python
# Create a copy for timestamp analysis
df_time = ambulances.copy()

# Convert timestamps to datetime
df_time["dispatch_dttm"] = pd.to_datetime(
    df_time["dispatch_dttm"], errors="coerce"
)

df_time["available_dttm"] = pd.to_datetime(
    df_time["available_dttm"], errors="coerce"
)

# Calculate occupation duration in minutes
df_time["busy_minutes"] = (
    df_time["available_dttm"] - df_time["dispatch_dttm"]
).dt.total_seconds() / 60

# Check for problems
print("Missing dispatch timestamps:",
      df_time["dispatch_dttm"].isna().sum())

print("Missing availability timestamps:",
      df_time["available_dttm"].isna().sum())

print("Negative durations:",
      (df_time["busy_minutes"] < 0).sum())

# Duration statistics by unit type
display(
    df_time.groupby("unit_type")["busy_minutes"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
)
```
